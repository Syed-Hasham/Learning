## Series
**1D LABELLED ARRAY CAPABLE OF HOLDING ANY DATATYPE**

```python
data = [101,102,103,201,202]
series = pd.Series(data,index = ["a","b","c","d","e"])
print(series)
```

```python
series.loc["c"] = 200
print(series.loc["c"])
```

```python
print(series.iloc[0]) #Basically works like accesing an array
```

```python
print(series[series > 200])
```

```python
calories = {"Day 1" : 1750, "Day 2" : 2100, "Day 3" : 1700}

series2 = pd.Series(calories)

print(series2)
```

```python
print(series2.loc["Day 1"])
```

```python
print(series2[series2>=2000]) #Filtering Calories Greater Than 2000
```

## DATAFRAMES
**Tabular Data Structure Containing Rows And Columns**

```python
data = {"Name": ["ABC","DEF","GHI"]
        ,"Age": [30, 35,45]
        }

df = pd.DataFrame(data, index = ["Employee 1", "Employee 2", "Employee 3"])

print(df)
print("\n")
```

```python
print(df.loc["Employee 3"])
```

```python
print(df.iloc[0])
```

```python
#Adding a new column
df["Job"] = ["XYZ", "UVW", "RST"]

print(df)
```

```python
#Adding a new row
row = pd.DataFrame([{"Name": "JKL", "Age": "50", "Job": "OPQ"},
                    {"Name": "MNO", "Age": "55", "Job": "LMN"},
                    {"Name": "PQR", "Age": "60", "Job": "IJK"}],
                   index = ["Employee 4","Employee 5","Employee 6"])
df = pd.concat([df,row])
print(df)
```

## Selection

```python
df = pd.read_csv("pokemon_data.csv",index_col="Name")
print(df.loc["Pikachu"])
```

## Filtering

```python
df = pd.read_csv("pokemon_data.csv")
tall_pokemon = df[df["Height"]>2]
print(tall_pokemon)
```

```python
legendary_pokemon = df[df["Legendary"]==1]
print(legendary_pokemon)
```

```python
water_pokemon = df[(df["Type1"] == "Water") | (df["Type2"] == "Water")]
print(water_pokemon)
```

```python
water_and_poison_pokemon = df[(df["Type1"]=="Water") & (df["Type2"]=="Poison")]
print(water_and_poison_pokemon)
```

## Aggregation
**Reduces a set of values into a single summary value**

```python
df = pd.read_csv("pokemon_data.csv")
print(df.mean(numeric_only=True))
```

```python
print(df.sum(numeric_only=True))
```

```python
print(df.min(numeric_only=True))
```

```python
print(df.max(numeric_only=True))
```

```python
print(df.count())
```

```python
#Selecting a single column
print(df["Height"].mean())
```

```python
group = df.groupby("Type1")
print(group["Height"].mean())
```

## Data Cleaning

```python
df = pd.read_csv("pokemon_data.csv")

#Drop Columns

df = df.drop(columns=["Legendary","No"])
print(df)
```

```python
df = pd.read_csv("pokemon_data.csv")

#Handle Missing Data

df = df.dropna(subset = ["Type2"])
print(df)
```

```python
df = pd.read_csv("pokemon_data.csv")

#Replace Missing Data

df = df.fillna({"Type2":"None"})
print(df)
```

```python
df = pd.read_csv("pokemon_data.csv")

#Fix Inconsistent Values

df["Type1"] = df["Type1"].replace({"Grass":"GRASS",
                                   "Fire":"FIRE",
                                   "Water":"WATER"})
print(df)
```

```python
df = pd.read_csv("pokemon_data.csv")

#Fix Datatypes

df["Legendary"] = df["Legendary"].astype(bool)
print(df)
```

```python
#This New Csv Contains Duplicate Values
df = pd.read_csv("pokemon_data_modified.csv")
print(df)
```

```python
#Remove Duplicate Values

df = df.drop_duplicates()
print(df)
```
