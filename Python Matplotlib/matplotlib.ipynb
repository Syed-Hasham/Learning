## Plot Customization

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array( [2023 , 2024 , 2025, 2026])
y1 = np.array([-10, -50, 0, -100])
y2 = np.array([-3, 50, 20, -10])

plt.plot(x, y1, marker = 'o', markersize = 10, markerfacecolor = 'cyan', linestyle = 'solid', color = 'r', linewidth = 3)
plt.plot(x, y2, marker = 'o', markersize = 10, markerfacecolor = 'magenta', linestyle = 'solid', color = 'b', linewidth = 3)
plt.xlabel('Year')
plt.ylabel('Financial Status')
plt.title('Financial Performance Over Time')
plt.show()
```

```python
# Another Way to pass the plot style as a dict
import matplotlib.pyplot as plt
import numpy as np
x = np.array( [2023 , 2024 , 2025, 2026])
y1 = np.array([10,20,40,100])
y2 = np.array([-3,-10,-20,-60])
line_style = dict(marker = 'o', markersize = 10, markerfacecolor = 'cyan', linestyle = 'solid', linewidth = 3)

plt.plot(x, y1, **line_style,color = "#18F114")
plt.plot(x,y2, color = "#d33f35",**line_style)
plt.xlabel('Year')
plt.ylabel('Financial Status')
plt.title('Financial Performance Over Time')
plt.show()  
```

## Labels

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class size", fontsize=25,
                        family="Arial",
                        fontweight="bold",
                        color="#2d4cfc")

plt.xlabel("Year", fontsize=20,
                   family="Arial",
                   fontweight="bold",
                   color="#2dbefc")
plt.tick_params(axis="both",colors = "#21331c")
plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)
plt.xticks(x)
plt.show()
```

## Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np
categories = np.array(["grains","fromage","meat","vegetables","fruits"])
categories = np.char.capitalize(categories)
values = np.array([20, 15, 30, 10, 25])
plt.bar(categories,values,color = "r")
#we can also use plt.barh for a vertical bar chart
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

## Pie Chart

```python
categories = ["Freshmen","Sophomore","Junior","Senior"]
values = [100, 150, 120, 80]
colors = ["#ff0000","#0080ff","#00ff00","#ff8000"]
plt.pie(values, 
        labels=categories, 
        autopct='%1.1f%%', 
        colors=colors,
        explode = (0,0.1,0,0),
        shadow = True,
        startangle = 90)
plt.title("Student Distribution")
plt.show()
```

## Scatter Plot

```python
import numpy as np
x1 = np.array([0 ,1 ,1 ,2, 3, 4, 5, 6, 7, 7, 8])
y1 = np.array([55, 68, 65, 62, 68, 78, 75, 78, 82, 85, 87])

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8])
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97])

plt.scatter(x1, y1,
            color = "red",
            alpha = 0.6,
            s = 20,
            label = "Class A")

plt.scatter(x2, y2,
            color = "blue",
            alpha = 0.6,
            s = 20,
            label = "Class B")


plt.xlabel("Hours Studied")
plt.ylabel("Grades")
plt.title("Study Time vs. Grades")
plt.legend()
plt.show()
```

## Histogram

```python
scores = np.random.normal(loc = 80 , scale = 10, size = 100)
scores = np.clip(scores, 0, 100)  # Ensure scores are between 0 and 100
plt.hist(scores)
plt.show()
```

## Subplots

```python
import matplotlib.pyplot as plt
import numpy as np
#print(plt.subplots(2, 3))

x1 = np.array([2023, 2024, 2025, 2026])

y1 = np.array([-10, -50, 0, -100])

y2 = np.array([10, 20, 40, 100])

y3 = np.array([5, 15, 25, 50])

y4 = np.array([-5, -20, -30, -60])

figure , axes =  plt.subplots(2, 2)

#we could also use bar chart, scatter plot, 
#histogram, pie chart, etc instead of line plot

axes[0, 0].plot(x1, y1)
axes[0, 0].set_title("Company A")

axes[0, 1].plot(x1, y2)
axes[0, 1].set_title("Company B")

axes[1,0].plot(x1, y3)
axes[1,0].set_title("Company C")

axes[1,1].plot(x1, y4)
axes[1,1].set_title("Company D")


plt.tight_layout()
plt.show()
```

## Pandas + Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("pokemon_data.csv")

type1_count = df["Type1"].value_counts(ascending=True) #Gives Count of coloumn Type1

plt.barh(type1_count.index, type1_count.values, color = "#BEA4F5", edgecolor = "black", linewidth = 1.5)
plt.title("Pokemon Type Distribution")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
plt.show()
```
